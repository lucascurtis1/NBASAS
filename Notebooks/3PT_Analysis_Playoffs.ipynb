```python
import pandas as pd
df_PV = pd.read_csv(r"C:\Users\skywa\OneDrive\Documents\FunCode\NBA3PT\Positional_Data\Point_Values_By_Season.csv")
```

```python
df_Wins25 = pd.read_csv(r"C:\Users\skywa\OneDrive\Documents\FunCode\NBA3PT\NBA_Team_Playoff_Stats\Team_Playoff_Wins_25.csv")
df_Wins24 = pd.read_csv(r"C:\Users\skywa\OneDrive\Documents\FunCode\NBA3PT\NBA_Team_Playoff_Stats\Team_Playoff_Wins_24.csv")
df_Wins23 = pd.read_csv(r"C:\Users\skywa\OneDrive\Documents\FunCode\NBA3PT\NBA_Team_Playoff_Stats\Team_Playoff_Wins_23.csv")
df_Wins22 = pd.read_csv(r"C:\Users\skywa\OneDrive\Documents\FunCode\NBA3PT\NBA_Team_Playoff_Stats\Team_Playoff_Wins_22.csv")
```

```python
print(df_PV)
print(df_Wins22)
```

```python
## Merging the data so that point value shows next to playoff wins.
```

```python
Merging_2022 = df_PV[["Team","2022"]].merge(
    df_Wins22,
    on="Team",
    how="inner"
)
Merging_2023 = df_PV[["Team","2023"]].merge(
    df_Wins22,
    on="Team",
    how="inner"
)
Merging_2024 = df_PV[["Team","2024"]].merge(
    df_Wins22,
    on="Team",
    how="inner"
)
Merging_2025 = df_PV[["Team","2025"]].merge(
    df_Wins22,
    on="Team",
    how="inner"
)
```

```python
print(Merging_2022)
print(Merging_2023)
print(Merging_2024)
print(Merging_2025)
```

```python
## Now I will perform a correlation test between playoff wins and Regular Season Thrpt shooting
```

```python
from scipy.stats import pearsonr
```

```python
r, p = pearsonr(
    Merging_2022["2022"],
    Merging_2022["W"]
)
print(f"r = {r}\np = {p}")
```

```python
r, p = pearsonr(
    Merging_2023["2023"],
    Merging_2023["W"]
)
print(f"r = {r}\np = {p}")
```

```python
r, p = pearsonr(
    Merging_2024["2024"],
    Merging_2024["W"]
)
print(f"r = {r}\np = {p}")
```

```python
r, p = pearsonr(
    Merging_2025["2025"],
    Merging_2025["W"]
)
print(f"r = {r}\np = {p}")
```

```python
## Now I have correlation stats for each of the 4 separate years. Next, I want to see what happens when I merge all of this data together and look at all 4 years together.
## Note: I see a positive correlation in all seasons, which is nice, however it is only statistically significant (at p < 0.05) in 2022 and 2024.
## Below, I am making sure the dataframes are ready for merge with common column names.
```

```python
df_2022 = Merging_2022.rename(columns={"2022": "PV"})
df_2022["Year"] = 2022

df_2023 = Merging_2023.rename(columns={"2023": "PV"})
df_2023["Year"] = 2023

df_2024 = Merging_2024.rename(columns={"2024": "PV"})
df_2024["Year"] = 2024

df_2025 = Merging_2025.rename(columns={"2025": "PV"})
df_2025["Year"] = 2025
```

```python
df_22_Through_25_Together = pd.concat(
    [df_2022,df_2023,df_2024,df_2025], ignore_index=True
)
```

```python
print(df_22_Through_25_Together)
```

```python
r, p = pearsonr(
    df_22_Through_25_Together["PV"],
    df_22_Through_25_Together["W"]
)
print(f"r = {r}\np = {p}")
```

```python
## The p value suggests very strong evidence that there is some nonzero correlation occuring.
## an r of 0.45 suggests that the correlation is positive, but it is only moderately strong.
## Now, I'd like to plot this in a scatterplot.
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(
    df_22_Through_25_Together["PV"],
    df_22_Through_25_Together["W"]
)

plt.xlabel("3PT Performance Score, Regular Season")
plt.ylabel("Playoff Wins")
plt.title("Regular Season 3PT Success vs. Subsequent Playoff Success")
plt.show()
```
